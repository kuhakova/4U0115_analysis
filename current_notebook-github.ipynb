```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.io.fits as fits
import warnings
```

# Defining the functions

```python
def cutting(time, rate, period, number, binsize, start=0, show='no', left=None, right=None, dotp=0, normalize=True, pl_style='normal', r_bounds=None):
    """
    Function for cutting light curve by pulses (if the pulsation period is known). Writes several values for each pulse (e.g. mean rate,
    start and stop time of each pulse etc). Helps to easily plot the light curve with different modifications.

    Keyword arguments:
    time -- numpy array with time in seconds. Time is expected to be without any NaNs. Can contain gaps.
    rate -- numpy array with rate. Can contain NaNs, in this case pulses containing NaNs are omitted and their index numbers are written
    in a list called "skipped".
    period -- float, period of pulsation in seconds, at the time of the observation start.
    number -- int, number of pulses to be cut. Must not be more than time length of the light curve divided by period pulsation.
    binsize -- float, time bin length in seconds for the light curve.
    start -- float, time from which the cutting process is started. Default is 0.
    show -- str, if you need to see or save the result. Default is "no" - no plot. Other options are: "show" - shows the plot or "save" - 
    saves the plot.
    left, right -- float, for specifying xlims for plot. By default left is stop time minus 90s and right is stop time + 10s.
    dotp -- float, period derivative. Default is 0. dotp can be used for calculating more precisely the pulses time of arrival.
    normalize -- boolean, whether the count rate needs to be normalized for each pulse. Default is True.
    pl_style -- str, how to plot the light curve: normal, colored plot (each pulse is colored corresponding to its count rate. Number of 
    groups for rate and consequently number of colors used is length of r_bounds+1), colored scatter (same as previous one but instead of 
    plotting whole pulse, one dot with value of mean rate of the pulse is plotted). Default is normal.
    r_bounds -- tuple, if colored plot or colored scatter is needed. Must specify boundaries in rate cate by which pulses are divided in 
    several groups and colored by different color. Rates in r_bounds must be placed in ascending order.

    Returns:
    I -- list with mean count rate for each pulse
    r -- list of numpy arrays with count rates for each pulse
    t_start, t_stop -- lists with time boundaries for each pulse
    skipped -- list index numbers of the skipped pulses (which have NaNs in rate)
    t0_i -- float, calculated stop time of the last pulse
    """
    
    # errors
    if (pl_style == 'colored plot' or pl_style == 'colored scatter') and r_bounds == None:
        raise ValueError('If you want colored plot or scatter, you need to specify r_bounds!')
    r_bounds = np.array(r_bounds)
    
    # all the needed lists:
    I = [] # mean intensity for each pulse
    r = [] # rate for each pulse
    t_start = [] # time boundaries of each pulse
    t_stop = []
    skipped = [] # index numbers of skipped pulses
    
    # making figure
    if show == 'show' or show == 'save':
        fig = plt.figure(figsize=(14,7))
        plt.xlabel('Time, s', fontsize=16)
        if normalize:
            plt.ylabel('Count rate, normalized units', fontsize=16)
        else:
            plt.ylabel('Count rate, counts/s', fontsize=16)
        plt.gca().tick_params(labelsize=14)

    # determining correct start time, as the entered value can fall into a gap in data
    start_ind = np.argmin(np.abs(time - start))
    t0 = time[start_ind] # t0 is the start time of the first pulse
    if np.abs(start - t0) >= binsize*1.1:
        warnings.warn(f'Warning! The entered start time falls into a gap. The correct start time, {t0}, will be used instead')
    t0_i = t0 # t0_i will be the start time of the current pulse in the following for cycle
    
    # the main cutting process
    for i in range(number):
        t_i = t0 + (i+1) * period + 1/2 * (i+1)**2 * period * dotp # next pulse time of arrival

        if t_i > time[-1]: # if the light curve has already ended, show a warning
            warnings.warn(f'Warning, number of pulses is too big! Currently light curve can be cut by {i} pulses.')
            
        stop_ind = np.argmin(np.abs(time - t_i))
        if (np.abs(time[stop_ind] - t_i) < binsize*1.1) and (all(np.diff(time[start_ind:stop_ind+1]) < binsize*1.1)): # checking for gaps
            I.append(np.mean(rate[start_ind:stop_ind])) # calculate mean intensity for the pulse
            r.append(rate[start_ind:stop_ind]) # write count rates of the current pulse
            t_start.append(time[start_ind]) # add time boundaries
            t_stop.append(time[stop_ind])

            if show == 'show' or show == 'save':
                if pl_style == 'normal':
                    if normalize:
                        plt.plot(time[start_ind:stop_ind], rate[start_ind:stop_ind] / I[-1], c='C0') # plot current pulse
                                                                                                     # with normalized rate
                        if i > number-10:
                            plt.scatter(time[stop_ind-1], rate[stop_ind-1] / I[-1], c='C3') # adding red dot at the pulse's end
                                                                                            # for more convenient look
                    else:
                        plt.plot(time[start_ind:stop_ind], rate[start_ind:stop_ind], c='C0') # plot current pulse
                        
                elif pl_style == 'colored plot':
                    if I[-1] >= r_bounds[-1]: # if mean rate of the current pulse is higher than the last rate boundary
                        c_number = str(len(r_bounds)) # color number
                        if normalize:
                            plt.plot(time[start_ind:stop_ind], rate[start_ind:stop_ind] / I[-1], c='C'+c_number)
                        else:
                            plt.plot(time[start_ind:stop_ind], rate[start_ind:stop_ind], c='C'+c_number)
                    else: # color number is defined from analysing in which boundaries the current pulse's mean rate lies:
                        c_number = str(np.where(I[-1] < r_bounds)[0][0]) # color number is defined from checking 
                                                                         # in which boundaries the current pulse's mean rate lies
                        if normalize:
                            plt.plot(time[start_ind:stop_ind], rate[start_ind:stop_ind] / I[-1], c='C'+c_number)
                        else:
                            plt.plot(time[start_ind:stop_ind], rate[start_ind:stop_ind], c='C'+c_number)
                            
                elif pl_style == 'colored scatter': # here normalization is not needed bc dots are plotted according 
                                                    # to the mean count rate of the pulse
                    if I[-1] >= r_bounds[-1]:
                        c_number = str(len(r_bounds))
                        plt.scatter((time[start_ind] + time[stop_ind]) / 2, I[-1], c='C'+c_number)
                    else:
                        c_number = str(np.where(I[-1] < r_bounds)[0][0])
                        plt.scatter((time[start_ind] + time[stop_ind]) / 2, I[-1], c='C'+c_number)
                        
            t0_i = t_i # going forward
            start_ind = stop_ind
            
        elif (np.abs(time[stop_ind] - t_i) < binsize*1.1) and (any(np.diff(time[start_ind:stop_ind+1]) >= binsize*1.1)): # if there's a gap
            skipped.append(i+1) # just skipping current pulse and going forward
            t0_i = t_i
            start_ind = stop_ind
        elif (np.abs(time[stop_ind] - t_i) >= binsize*1.1): # if the stop time falls into a gap
            skipped.append(i+1) # skipping current pulse, but the start index of the next pulse is unknown yet, it will be calculated
            t0_i = t_i          # durnig next interation
        else:
            raise ValueError(f'Something has not gotten to any elifs in the main cutting process on the step {i}')

    if show == 'save' or show == 'show': # finishing the plot
        if (left != None) & (right != None):
            plt.xlim(left, right) # add specified xlims
        else:
            plt.xlim(time[stop_ind]-90, time[stop_ind]+10) # or the default ones
        if show == 'save':
            plt.savefig('lc normed ' + str(left)[:4]+'.png', dpi=300) # saving
        else:
            plt.show() # showing
    
    return np.array(I), r, np.array(t_start), np.array(t_stop), skipped, t0_i
```

```python
def efoldin(I, rate, n, bins=36, show='no', normalization='mean'):
    """
    Function for making pulse profiles for several quantiles of the pulses distribution. Helps to easily plot the profiles.
    
    Keyword arguments:
    I -- list or array with mean intensities of pulses.
    rate -- list or array with count rates for each pulse. The pulses can have slightly different "length" (e.g. 35, 36, 37).
    n -- int, number of quantiles or parts for which the pulses distribution will be equally sliced by. n = 1 plots mean pulse profile for 
    the whole data.
    bins -- int, number of bins for the pulse profiles. Default is 36.
    show -- str, if you need to see or save the result. Default is "no" - no plot. Other options are: "show" - shows the plot or "save" - 
    saves the plot.
    normalization -- str, how to normalize pulse profiles. Default is "mean" - each pulse profile is normalized by its mean value. Other
    options are: "max" -- each pulse is normalized by its maximum value.

    Returns:
    x -- numpy array, linspace of phases from 0 to 1 with the specified number of bins.
    profiles -- list of all the pulse profiles in ascending order by mean intensity. Without normalization by mean or max.
    profiles_normed -- list of all the pulse profiles with specified normalization.
    """

    # defining lists
    profiles = []
    profiles_normed = []
    # making figure
    if show == 'show' or show == 'save':
        fig = plt.figure(figsize=(12, 7))
        plt.xlabel('Phase', fontsize=16)
        plt.ylabel('Count rate, normalized units', fontsize=16)
        plt.gca().tick_params(labelsize=14)
    
    # defining percentile
    I_sorted = np.sort(I)
    percentile = []
    for i in I_sorted:
        percentile.append(np.sum(np.where(I_sorted >= i, 1, 0)) / I_sorted.shape[0])
    percentile = np.array(percentile)
    
    # making linspace with phases
    x = np.linspace(0, 1-1/bins, bins)

    # the main process
    min_intensity = 0 # min intensity for the first part
    for i in range(n):
        profile_i = np.zeros(bins) # current pulse profile
        num = 0 # number of pulses used to make current pulse profile
        max_intensity = I_sorted[percentile >= 1 - (i+1)/n][-1] # calculate the upper intensity boundary for the current part
        # adding pulses to current pulse profile
        for j in range(len(I)):
            if I[j] >= min_intensity and I[j] <= max_intensity: # if intensity is in the desired range
                if len(rate[j]) == bins: # if current pulse is binned by the needed number of bins -> simply add it to profile
                    profile_i += rate[j]
                else: # if not -> use interpolation
                    xp = np.linspace(0, 1-1/len(rate[j]), len(rate[j])) # phases of the current pulse
                    fp = rate[j]
                    profile_i += np.interp(x, xp, fp)
                num += 1 
                
        profile_i /= num # normalize current pulse profile by the number of pulses taken into account
        profiles.append(profile_i)
        # normalization:
        if normalization == 'mean':
            profile_i_normed = profile_i / np.mean(profile_i)
        elif normalization == 'max':
            profile_i_normed = profile_i / np.max(profile_i)
        else:
            print('''The value passed to "normalization" argument is incorrect! The possible options are: mean and max. Currently the 
            default (mean) normalization will be used.''')
            profile_i_normed = profile_i / np.mean(profile_i)
        profiles_normed.append(profile_i_normed)

        if show == 'show' or show == 'save':
            plt.plot(x, profile_i_normed)
        if i < n - 1:
            min_intensity = I_sorted[percentile < 1 - (i+1)/n][0]

    # finishing plot
    if show == 'save':
        plt.savefig('pulse profile ' + str(n) + ' quantiles.png', dpi=300) # saving
    else:
        plt.show() # showing

    return x, profiles, profiles_normed
```

```python
def percentile(a):
    b = np.sort(np.array(a))
    perc = []
    for i in b:
        perc.append(np.sum(np.where(b >= i, 1, 0)) / b.shape[0])
    return np.array(perc)
```
